## **Producing Visualizations**

```python
#Importing
import numpy as np
import pandas as pd
```

```python
#2005-06 Data
#Prices
ugd4_p0 = pd.read_csv('Data Files/Uganda - Food Prices (2005-06).csv')
ugd4_p0 = ugd4_p0[ugd4_p0["u"] == "Kg"]
ugd4_p = pd.pivot_table(ugd4_p0, values="price", index=["j","u"], columns=["t","m"] , aggfunc="median")
#Food Expenditures
ugd4_x0 = pd.read_csv('Data Files/Uganda - Food Expenditures (2005-06).csv')
ugd4_x0['i'] = ugd4_x0['i'].astype(str)
ugd4_x = pd.pivot_table(ugd4_x0, values="Expenditure", index=["i","t","m"], columns="j" , aggfunc="median")
#Consumption
ugd4_x_long = ugd4_x.stack().reset_index(name="Expenditure")
ugd4_p_long = ugd4_p0
ugd4_c = ugd4_x_long.merge(ugd4_p_long, on=["j", "t", "m"], how="left")
ugd4_c["Consumption"] = ugd4_c["Expenditure"] / ugd4_c["price"]
ugd4_c = ugd4_c.pivot_table(values="Consumption", index=["i","t","m","u"], columns="j", aggfunc="sum")
#HH Characteristics
ugd_z0 = pd.read_csv('Data Files/Uganda - Household Characteristics.csv')
ugd4_z = ugd_z0[ugd_z0['t'] == '2005-06']
ugd4_z = ugd4_z.set_index(['i','t','m'])
ugd4_z.columns.name = 'k'
ugd4_z.fillna(0, inplace=True)
```

### Price Trends Over Time

```python
# List of datasets
price_dfs = [ugd4_p, ugd2_p, ugd5_p, ugd3_p, ugd6_p, ugd_p, ugd7_p, ugd8_p]
years = ['2005-06', '2009-10', '2010-11', '2011-12', '2013-14', '2015-16', '2018-19', '2019-20']

# Aggregate prices for selected staple foods
staple_foods = ['Matoke', 'Sweet Potatoes (fresh)', 'Cassava (fresh)', 'Cassava (dry)', 'Maize (flour)']

# Extract median prices over time
price_data = []
for df, year in zip(price_dfs, years):
    for food in staple_foods:
        if food in df.index.get_level_values('j'):
            median_price = df.loc[food].median().median()
            price_data.append({'Year': year, 'Food': food, 'Price': median_price})

# Convert to DataFrame
price_df = pd.DataFrame(price_data)

# Plot
plt.figure(figsize=(10,6))
sns.lineplot(data=price_df, x='Year', y='Price', hue='Food', marker='o')
plt.title('Price Trends of Staple Foods Over Time')
plt.xlabel('Year')
plt.ylabel('Median Price (UGX per Kg)')
plt.xticks(rotation=45)
plt.legend(title="Food Item")
plt.show()
```

### Inflation-Adjusted Prices

```python
# Normalize prices to 2005-06 as base (set it to 100)
base_prices = price_df[price_df['Year'] == '2005-06'].set_index('Food')['Price']
price_df['Inflation Adjusted Price'] = price_df.apply(lambda row: (row['Price'] / base_prices[row['Food']]) * 100 if row['Food'] in base_prices else None, axis=1)

# Plot
plt.figure(figsize=(10,6))
sns.lineplot(data=price_df, x='Year', y='Inflation Adjusted Price', hue='Food', marker='o')
plt.title('Inflation-Adjusted Prices of Staple Foods Over Time')
plt.xlabel('Year')
plt.ylabel('Price Index (Base Year = 100)')
plt.xticks(rotation=45)
plt.legend(title="Food Item")
plt.show()
```

### Expenditure Breakdown by Category

```python
expenditure_dfs = [ugd4_x, ugd2_x, ugd5_x, ugd3_x, ugd6_x, ugd_x, ugd7_x, ugd8_x]
category_data = []

# Filter expenditure data for only staple foods and aggregate
for df, year in zip(expenditure_dfs, years):
    df_filtered = df[staple_foods]  # Keep only the staple foods
    category_sums = df_filtered.sum().reset_index()
    category_sums.columns = ['Food', 'Expenditure']
    category_sums['Year'] = year
    category_data.append(category_sums)

# Combine all years into one DataFrame
expenditure_df = pd.concat(category_data)

# Plot
plt.figure(figsize=(12,6))
sns.barplot(data=expenditure_df, x='Year', y='Expenditure', hue='Food')
plt.title('Food Expenditure Breakdown Over Time (Staple Foods)')
plt.xlabel('Year')
plt.ylabel('Total Expenditure (UGX)')
plt.xticks(rotation=45)
plt.legend(title="Food Category", bbox_to_anchor=(1,1))
plt.show()
```

```python
# Merge consumption with FCT to get nutrient intake
nutrient_intake = ugd_fct.multiply(ugd4_c, axis=0)  # Example for 2005-06
nutrient_intake = nutrient_intake.groupby(level=0).sum()  # Sum across households

# Compare against RDI
rdi_values = ugd_rdi.mean(axis=1)  # Average RDI for all categories
sufficiency = (nutrient_intake / rdi_values) * 100

# Reshape for visualization
sufficiency_long = sufficiency.stack().reset_index()
sufficiency_long.columns = ['Food', 'Nutrient', 'Percentage of RDI Met']

# Plot
plt.figure(figsize=(12,6))
sns.heatmap(sufficiency.pivot(index='Food', columns='Nutrient', values='Percentage of RDI Met'), cmap='coolwarm', annot=True)
plt.title('Nutrient Sufficiency (2005-06)')
plt.xlabel('Nutrient')
plt.ylabel('Food Item')
plt.show()
```
